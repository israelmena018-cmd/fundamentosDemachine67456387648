## 1. Marco teórico: clasificación en modelos de predicción de IA

### 1.1 ¿Qué es la clasificación?
La **clasificación** es una tarea de **aprendizaje supervisado**: a partir de ejemplos
etiquetados `(X, y)` el modelo aprende a asignar a una observación nueva una **categoría
discreta** (clase). Se diferencia de la **regresión**, donde la variable objetivo es continua.

| Tipo | Descripción | Ejemplo |
|---|---|---|
| **Binaria** | Dos clases posibles | Correo spam / no spam |
| **Multiclase** | Más de dos clases, una sola por muestra | Especie de flor Iris (3 clases) |
| **Multietiqueta** | Cada muestra puede tener varias etiquetas a la vez | Etiquetar una foto con "playa", "persona", "atardecer" |

### 1.2 Algoritmos habituales
Regresión logística, k-NN, árboles de decisión, Random Forest, Naive Bayes,
**SVM (Support Vector Machine)**, gradient boosting y redes neuronales.

### 1.3 Flujo de trabajo de un problema de clasificación
1. **Cargar los datos** y separar características `X` y objetivo `y`.
2. **Partición externa (hold-out)**: separar un conjunto de **test** que no se toca hasta el final.
3. **Preprocesar** (p. ej. estandarizar) ajustando solo con los datos de entrenamiento.
4. **Validación interna** (cross-validation) para evaluar y elegir hiperparámetros.
5. **Entrenar el modelo definitivo** con todo el conjunto de entrenamiento.
6. **Evaluar una única vez en test** para estimar el rendimiento real.

### 1.4 Hold-out y validación cruzada (cross-validation)
- **Hold-out**: se reserva un porcentaje (aquí 20 %) para test. Simula datos nunca vistos.
- **K-fold cross-validation**: el conjunto de entrenamiento se divide en *k* bolsas (*folds*).
  Se entrena con *k-1* y se valida con la restante, rotando *k* veces. Así cada dato se usa
  para validar exactamente una vez, y se obtiene una estimación más estable.
- `cross_val_predict` devuelve la predicción de cada muestra cuando estuvo en la bolsa de validación.

### 1.5 Estandarización
`StandardScaler` transforma cada variable a media 0 y desviación 1: $z = (x - \mu)/\sigma$.
Es importante en SVM, k-NN y redes neuronales porque son sensibles a la escala.
Se **ajusta (`fit`) solo con train** y se **aplica (`transform`) a test** para evitar *data leakage*
(fuga de información).

### 1.6 SVM con kernel RBF
Una SVM busca el **hiperplano** que separa las clases con el mayor margen. Con el kernel **RBF**
puede crear fronteras no lineales. Hiperparámetros clave:
- **C**: penalización por errores. C pequeño → margen amplio, modelo más simple (riesgo de *underfitting*);
  C grande → ajusta más a los datos (riesgo de *overfitting*).
- **gamma**: alcance de influencia de cada punto. Gamma alto → fronteras muy ajustadas; gamma bajo → más suaves.

### 1.7 Matriz de confusión y métricas
Para un problema binario, la matriz de confusión contiene:

| | Predicho positivo | Predicho negativo |
|---|---|---|
| **Real positivo** | VP (verdadero positivo) | FN (falso negativo) |
| **Real negativo** | FP (falso positivo) | VN (verdadero negativo) |

| Métrica | Fórmula | Interpretación |
|---|---|---|
| **Exactitud (accuracy)** | $(VP+VN)/(VP+VN+FP+FN)$ | Proporción total de aciertos. Engañosa con clases desbalanceadas |
| **Precisión (precision)** | $VP/(VP+FP)$ | De lo que predije positivo, cuánto lo era realmente |
| **Sensibilidad (recall)** | $VP/(VP+FN)$ | De los positivos reales, cuántos detecté |
| **F1-score** | $2\cdot\dfrac{P\cdot R}{P+R}$ | Media armónica de precisión y sensibilidad |

En **multiclase** se calcula cada métrica por clase y luego se promedia:
- `macro`: promedio simple entre clases (todas pesan igual).
- `weighted`: promedio ponderado por el número de muestras de cada clase.
- `micro`: cuenta VP, FP y FN globalmente.

### 1.8 Sobreajuste (overfitting) vs. subajuste (underfitting)
- **Overfitting**: muy buen resultado en train, malo en test (el modelo memoriza).
- **Underfitting**: mal resultado incluso en train (el modelo es demasiado simple).
Comparar el rendimiento de validación con el de test ayuda a detectarlos.

## 2. Importación de librerías

```python
import numpy as np                                   # Operaciones numéricas con arreglos
from sklearn import datasets                         # Conjuntos de datos de ejemplo (Iris)
from sklearn.dummy import DummyClassifier            # Clasificador base de referencia (baseline)
from sklearn.model_selection import cross_val_predict, cross_validate, cross_val_score  # Validación cruzada
from sklearn.model_selection import KFold            # Generador de particiones k-fold
from sklearn.model_selection import GridSearchCV     # Búsqueda de hiperparámetros con validación cruzada
import sklearn.metrics as metrics                    # Métricas de evaluación
from sklearn.model_selection import train_test_split # Partición train/test (hold-out)
from sklearn import preprocessing                    # Estandarización de datos
from pprint import pprint                            # Impresión legible de objetos
from sklearn.svm import SVC                          # Support Vector Classifier
```

## 3. Carga de datos
El dataset **Iris** tiene 150 flores de 3 especies (*setosa, versicolor, virginica*), descritas por
4 características: largo y ancho del sépalo y del pétalo (en cm). Es un problema de **clasificación multiclase**.

```python
# Carga de datos.
datos = datasets.load_iris()
print(np.shape(datos.data))   # (150 muestras, 4 características)
```

```python
# Extraer la matriz de datos "X" (características) y la variable target "y" (clase)
X = datos.data
y = datos.target
print("Clases:", datos.target_names)
print("Muestras por clase:", np.bincount(y))   # Iris está balanceado: 50 por clase
```

## 4. Partición externa (hold-out 80-20 %)
Reservamos el 20 % como **test**. No se usará hasta la evaluación final.
`random_state=42` hace el resultado reproducible y `stratify=y` mantendría la proporción de clases
(opcional; aquí se conserva el planteamiento original).

```python
# Partición EXTERNA. Test: hold-out split 80-20%.
X_train, X_test, y_train, y_test = train_test_split(datos.data, datos.target, test_size=0.2, random_state=42)
print('Train dimensions: ', np.shape(X_train))
print('Test dimensions:  ', np.shape(X_test))
```

## 5. Estandarización
Se ajusta el escalador **solo con train** (`fit_transform`) y se aplica a test con los mismos
parámetros (`transform`), para no filtrar información del test al entrenamiento.

```python
# Estandarización de los datos de entrenamiento y test
standardizer = preprocessing.StandardScaler()
X_train = standardizer.fit_transform(X_train)   # Calcula media y desviación con train y transforma
X_test = standardizer.transform(X_test)         # Aplica los mismos parámetros a test
```

## 6. Definición del algoritmo
Usamos una **SVM con kernel RBF**. `C=0.1` es una penalización baja (modelo suave) y
`gamma='auto'` equivale a $1/n_{caracteristicas}$.

```python
# Definimos el algoritmo
alg = SVC(C=0.1, gamma='auto', kernel='rbf')
```

## 7. Validación cruzada interna (5 folds)
`cross_val_predict` entrena y valida 5 veces sobre el conjunto de **train** y devuelve la predicción
de cada muestra cuando estuvo en la bolsa de validación.

> **Corrección del error original:** antes se escribía `y_pred = cross_val_predict` (sin paréntesis ni argumentos),
> lo que guardaba la *función* en lugar de las predicciones y provocó el `InvalidParameterError`.
> Hay que **llamar** a la función pasándole el modelo, los datos, las etiquetas y el número de folds.

```python
# Cross-validation interno
# Utilizar comando "cross_val_predict" para extraer directamente las predicciones de las 5 bolsas de validación
y_pred = cross_val_predict(alg, X_train, y_train, cv=5)   # cv=5 -> 5 bolsas
print("Primeras 20 predicciones:", y_pred[:20])
print("Primeras 20 reales:      ", y_train[:20])
```

## 8. Resultados de la validación
Como hay 3 clases usamos `average="macro"`: se calcula la métrica de cada clase y se promedia con el mismo peso.

```python
# Presentación de los resultados de la evaluación.
print("Exactitud: %.3f" % (metrics.accuracy_score(y_train, y_pred)))                      # accuracy
print("Precisión: %.3f" % (metrics.precision_score(y_train, y_pred, average="macro")))    # precision
print("Sensibilidad: %.3f" % (metrics.recall_score(y_train, y_pred, average="macro")))    # sensibilidad (recall)
print("F1-score: %.3f" % (metrics.f1_score(y_train, y_pred, average="macro")))            # F-score
```

### Matriz de confusión
Las **filas** son las clases reales y las **columnas** las predichas. La diagonal principal son los aciertos;
fuera de la diagonal están los errores (qué clase se confunde con cuál).

```python
# Extraemos la matriz de confusión
print("Matriz de confusión:\n", metrics.confusion_matrix(y_train, y_pred))
```

```python
# Informe con precisión, recall, F1 y soporte (nº de muestras) por clase
print("Tabla de métricas:\n", metrics.classification_report(y_train, y_pred, target_names=datos.target_names))
```

## 9. Optimización de hiperparámetros (GridSearchCV)
El enunciado habla de "hiperparámetros óptimos". Para obtenerlos probamos varias combinaciones de
`C` y `gamma` con validación cruzada **solo sobre train**, y nos quedamos con la mejor.

```python
# Rejilla de hiperparámetros a probar
param_grid = {'C': [0.01, 0.1, 1, 10, 100],
              'gamma': ['auto', 0.01, 0.1, 1]}

# Búsqueda con 5-fold CV sobre train (el test sigue sin tocarse)
grid = GridSearchCV(SVC(kernel='rbf'), param_grid, cv=5, scoring='accuracy')
grid.fit(X_train, y_train)

print("Mejores hiperparámetros:", grid.best_params_)
print("Mejor exactitud en validación: %.3f" % grid.best_score_)
```

## 10. Modelo definitivo y evaluación en test
Con los hiperparámetros óptimos, entrenamos con **todos los datos de train** (en la validación cruzada
cada modelo solo veía 4/5 de ellos) y evaluamos **una sola vez** en test con `score()`, que devuelve la exactitud.

```python
# Una vez hemos optimizado los hiperparámetros con el conjunto de validación, utilizamos todos los datos de
# "train" y "validation" para entrenar un modelo definitivo con esos hiperparámetros ajustados
model = SVC(kernel='rbf', **grid.best_params_)   # Definimos el modelo con los hiperparámetros óptimos
model = model.fit(X_train, y_train)              # Entrenamos el modelo con todos los datos de "train" + "val"

# y finalmente evaluamos el test con el método "score()"
test_results = model.score(X_test, y_test)
print('Exactitud en test: ', np.round(test_results*100,4), '%')
```

```python
# obtenemos las predicciones del test para extraer el resto de métricas de interés
y_pred_test = model.predict(X_test)                                                          # Extraer predicciones
print(metrics.classification_report(y_test, y_pred_test, target_names=datos.target_names))   # Aplicar el método de classification_report()
print(metrics.confusion_matrix(y_test, y_pred_test))                                         # Extraer la matriz de confusión
```

## 11. Conclusiones
- La **validación cruzada** estima el rendimiento sin gastar el conjunto de test, y permite elegir hiperparámetros.
- El **test** se usa una sola vez al final: es la mejor estimación del comportamiento con datos nuevos.
- Si la exactitud de validación y la de test son parecidas, el modelo **generaliza bien**; una gran diferencia indicaría sobreajuste.
- Con Iris (clases balanceadas) la exactitud es una métrica válida, pero con clases desbalanceadas conviene
  mirar **precisión, recall y F1** por clase, y la matriz de confusión.
- **Mejora posible:** usar un `Pipeline(StandardScaler, SVC)` dentro de la validación cruzada para que el escalador
  se ajuste solo con las bolsas de entrenamiento de cada fold y evitar una pequeña fuga de información.
